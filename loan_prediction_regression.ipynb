```python
def remove_front_space(x):
  if x[0] == " ":
    return x[1:]
  return x
```

```python
remove_front_space(" hello")
```

```python
df["education"] = df["education"].apply(remove_front_space)
df["loan_status"] = df["loan_status"].apply(remove_front_space)
df["self_employed"] = df["self_employed"].apply(remove_front_space)
```

```python
def transform_education(x):
  if x == "Graduate":
    return 1
  if x == "Not Graduate":
    return 0

  raise Exception("Invalid value")


transform_education("Not Graduate")
```

```python
def transform_self_employed(x):
  if x == "Yes":
    return 1
  if x == "No":
    return 0

  raise Exception("Invalid value Yes or No expected")

transform_self_employed("Yes")
```

```python
df["education"] = df["education"].apply(transform_education)
df["self_employed"] = df["self_employed"].apply(transform_self_employed)
df.head()
```

```python
df = df[df["loan_status"] == 'Approved']
df
```

```python
df.shape
```

```python
df = df.drop(["loan_id", "loan_status"] , axis=1 , errors="ignore")
```

```python
#train test split
from sklearn.model_selection import train_test_split

X = df.drop('loan_amount', axis=1)
y = df['loan_amount']

# Perform train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
df
```

```python
#fit a regression model

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

```python
import statsmodels.api as sm

# Add a constant to the independent variables for statsmodels for training
X_train_sm = sm.add_constant(X_train)

# Fit the OLS model using statsmodels (this is needed for get_prediction)
ols_model = sm.OLS(y_train, X_train_sm).fit()

# Prepare X_test[:1] for prediction with statsmodels, ensuring a constant is added
# and column order matches X_train_sm.
# First, convert X_test[:1] to a DataFrame if it isn't already (though it usually is).
X_test_for_ols = X_test[:1].copy()

# Explicitly add the constant column. This is more robust than sm.add_constant on its own
# if the DataFrame structure or kernel state is causing issues.
X_test_for_ols.insert(0, 'const', 1.0)

# Ensure the column order of the prediction input matches the training data.
# This is critical for statsmodels' get_prediction method.
X_test_for_ols = X_test_for_ols[X_train_sm.columns]

# Get the predicted value and 95% confidence interval using the OLS model
predictions_summary = ols_model.get_prediction(X_test_for_ols).summary_frame(alpha=0.05)

# The predicted value from the sklearn model is already in the 'predicted' variable.
# The 'mean' from statsmodels.get_prediction should be very close.
predicted_value_ols = predictions_summary['mean'].iloc[0]
lower_bound = predictions_summary['obs_ci_lower'].iloc[0]
upper_bound = predictions_summary['obs_ci_upper'].iloc[0]

print(f"Predicted value (from sklearn model): {predicted[0]:.2f}")
print(f"Predicted value (from OLS model for interval calculation): {predicted_value_ols:.2f}")
print(f"95% Confidence Interval for the prediction: [{lower_bound:.2f}, {upper_bound:.2f}]")
```
